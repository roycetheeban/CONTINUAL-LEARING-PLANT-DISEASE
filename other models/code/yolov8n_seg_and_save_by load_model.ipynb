```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# =============================================
# 1. Configuration
# =============================================
# Define your paths
dataset_path = '/content/drive/MyDrive/yolo_seg_input'
model_save_dir = '/content/drive/MyDrive/seg_model/seg_model'
test_images_dir = f'{dataset_path}/yolo_seg_input'
output_dir = f'{dataset_path}/output'

# =============================================
# 2. Segmentation with Color Preservation
# =============================================
def segment_leaf(image_path, model_path):
    """Segment leaf while preserving original colors perfectly"""
    # Load trained model
    model = YOLO(model_path)

    # Run prediction
    results = model.predict(image_path, conf=0.7)

    for result in results:
        # Read original image (BGR format)
        orig_img = cv2.imread(image_path)

        if result.masks is not None:
            # Get best mask (largest area)
            mask = result.masks[np.argmax([np.sum(m.data.cpu().numpy()) for m in result.masks])]
            mask = mask.data[0].cpu().numpy()
            mask = cv2.resize(mask, (orig_img.shape[1], orig_img.shape[0]))
            mask = (mask > 0.5).astype(np.uint8)

            # Create white background output
            white_bg = np.ones_like(orig_img) * 255
            output_img = np.where(mask[..., None], orig_img, white_bg)

            # Visualization
            plt.figure(figsize=(15, 5))

            plt.subplot(1, 3, 1)
            plt.imshow(cv2.cvtColor(orig_img, cv2.COLOR_BGR2RGB))
            plt.title("Original Image")
            plt.axis('off')

            plt.subplot(1, 3, 2)
            plt.imshow(mask, cmap='gray')
            plt.title("Segmentation Mask")
            plt.axis('off')

            plt.subplot(1, 3, 3)
            plt.imshow(cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB))
            plt.title("Segmented (White BG)")
            plt.axis('off')

            plt.tight_layout()
            plt.show()

            print(f"✅ Processed: {os.path.basename(image_path)}")
        else:
            print(f"❌ No leaf detected in {os.path.basename(image_path)}")

# =============================================
# 3. Main Execution
# =============================================
if __name__ == "__main__":
    # Create directories if needed
    os.makedirs(output_dir, exist_ok=True)
    os.makedirs(model_save_dir, exist_ok=True)

    # Model path (using your existing best model)
    best_model_path = f'{model_save_dir}/middle_leaf_segmentation_best.pt'

    if not os.path.exists(best_model_path):
        raise FileNotFoundError(f"Model not found at {best_model_path}")

    # Process test images
    if os.path.exists(test_images_dir):
        test_images = sorted([f for f in os.listdir(test_images_dir)
                          if f.lower().endswith(('.jpg', '.jpeg', '.png'))])[:5]  # First 5 images

        print(f"🔍 Found {len(test_images)} test images")

        for img_name in test_images:
            img_path = os.path.join(test_images_dir, img_name)
            print(f"\nProcessing: {img_name}")
            segment_leaf(img_path, best_model_path)
    else:
        print(f"❌ Test directory not found: {test_images_dir}")
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# =============================================
# 1. Configuration
# =============================================
# Define your paths
dataset_path = '/content/drive/MyDrive/yolo_seg_input'
model_save_dir = '/content/drive/MyDrive/seg_model/seg_model'
test_images_dir = f'{dataset_path}/yolo_seg_input'
output_dir = os.path.join(test_images_dir, 'segmented_output')  # Create output dir inside input dir

# =============================================
# 2. Segmentation with Color Preservation
# =============================================
def segment_leaf(image_path, model_path, save_output=True):
    """Segment leaf while preserving original colors perfectly"""
    # Load trained model
    model = YOLO(model_path)

    # Run prediction
    results = model.predict(image_path, conf=0.7)

    for result in results:
        # Read original image (BGR format)
        orig_img = cv2.imread(image_path)

        if result.masks is not None:
            # Get best mask (largest area)
            mask = result.masks[np.argmax([np.sum(m.data.cpu().numpy()) for m in result.masks])]
            mask = mask.data[0].cpu().numpy()
            mask = cv2.resize(mask, (orig_img.shape[1], orig_img.shape[0]))
            mask = (mask > 0.5).astype(np.uint8)

            # Create white background output
            white_bg = np.ones_like(orig_img) * 255
            output_img = np.where(mask[..., None], orig_img, white_bg)

            # Visualization
            plt.figure(figsize=(15, 5))

            plt.subplot(1, 3, 1)
            plt.imshow(cv2.cvtColor(orig_img, cv2.COLOR_BGR2RGB))
            plt.title("Original Image")
            plt.axis('off')

            plt.subplot(1, 3, 2)
            plt.imshow(mask, cmap='gray')
            plt.title("Segmentation Mask")
            plt.axis('off')

            plt.subplot(1, 3, 3)
            plt.imshow(cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB))
            plt.title("Segmented (White BG)")
            plt.axis('off')

            plt.tight_layout()
            plt.show()

            # Save only the processed output image
            if save_output:
                os.makedirs(output_dir, exist_ok=True)
                output_filename = os.path.join(output_dir, f"segmented_{os.path.basename(image_path)}")
                cv2.imwrite(output_filename, output_img)
                print(f"💾 Saved output to: {output_filename}")

            print(f"✅ Processed: {os.path.basename(image_path)}")
        else:
            print(f"❌ No leaf detected in {os.path.basename(image_path)}")

# =============================================
# 3. Main Execution
# =============================================
if __name__ == "__main__":
    # Model path (using your existing best model)
    best_model_path = f'{model_save_dir}/middle_leaf_segmentation_best.pt'

    if not os.path.exists(best_model_path):
        raise FileNotFoundError(f"Model not found at {best_model_path}")

    # Process test images
    if os.path.exists(test_images_dir):
        test_images = sorted([f for f in os.listdir(test_images_dir)
                          if f.lower().endswith(('.jpg', '.jpeg', '.png')) and not f.startswith('segmented_')])[:14]  # First 14 images

        print(f"🔍 Found {len(test_images)} test images")
        print(f"📂 Output will be saved to: {output_dir}")

        for img_name in test_images:
            img_path = os.path.join(test_images_dir, img_name)
            print(f"\nProcessing: {img_name}")
            segment_leaf(img_path, best_model_path, save_output=True)
```
